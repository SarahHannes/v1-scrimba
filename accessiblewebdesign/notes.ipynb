# [1. Understanding Web Accessibility](#1-understanding-web-accessibility)

ATs - Assistive Technologies:
- Screen readers:
    - most common 
    - rely heavily on headings, links and form lavels for navigation.
- Speech recognition software
- Screen magnifiers
- Alternative input devices: designed for people with motor skill impairments.

- Accessibility tree: a key part of how browsers interact with assistive technology.
    - The browser translate these codes into visual UI; and at the same time it translate our code into an object called the accessibility tree.
    - Accessibility tree contains a lot of important information and context in text form.
    - The information inside this accessibility tree provide assistive technologies with this important context.

> 

- The below elements look the same because we styled them the same way; but in the accessibility tree, the button made using the `<div>` tag lacks information and context.
- When you use semantic HTML (using the right element for the right purpose), the browser automatically feeds the correct information in to the accessibility tree.
    - This helps assistive technologies to navigate and interprete the content accurately.
    - By using the correct tags (eg button tag, a tag, form controls), you get keyboard accessibility for free. This means these elements are naturally focusable and operable with a keyboard with no extra coding required.

# [3. Accessibility Guidelines](#3-accessibility-guidelines)

- Web Content Accessibility Guidelines (WCAG) 2.0 has different labels:
    - Level A: not accessible
    - Level AA: accessible
    - Level AAA: highest accessibility


# [4. Color contrast](#4-color-contrast)

- The color of text against its background color.
- We can check contrast ratio using online contrast checker, eg https://webaim.org/resources/contrastchecker/ or on the devtools
- To improve contrast ratio, we can do:
    - change the text color
    - change the background color (eg make it brighter using `background-blend-mode`)
    - make the text bigger

- Instead of using color alone, it is better to use color with text and shapes.

# [5. Alternative text](#5-alternative-text)

- Use conscise alt text.
- No need to start with the sentence of "an image of..." because screen readers will state that it is an image anyway.
- No need to add alt text to image that is purely for decoration.
- Why we need to add alt text?
    - helps people with little or no vision
    - helps people who have turned off images eg to save data
    - search engines like Google to find appropriate content

- End alt text with a period `.` to make a screen reader briefly pause, which can be helpful.
- When there is text inside an image (eg a cartoon image with dialogue bubble and texts), screen reader wont read it, so we can add this information in our alt text.
- Without alt text, the screen reader will read image filename instead, and that can be confusing/ annoying.

> 

# [6. Links](#6-links)

- Should use `<a>` element for links.
- Links should be recognizable as links.
    - Eg different color than the rest of the text, underline, make it bigger / bolder.
- Links should have non-ambiguous text.
    - Example of bad practice are "Click here", "More", "Continue" etc.
    - Make sure the text could be understand even out of context (eg people reading only the link text can identify that it is a link, and what will happen when they clicking on it.)
