# [1. Class in Javascript](#1-class-in-javascript)

- A "class" in JavaScript is a blueproint for creating object that have the same properties and access to the same methods.
- Class is just a "syntactic sugar" on top of setting up prototype manually
- We can use "class field" to directly initialize every object created from that class with the same value (no need to use constructor)
- If we need to initialize dynamic object properties, we can use `constructor()` method inside that class.
- You can "extend" from one class to another class to inherit all methods defined in the parent classes.

# [5. Lifecycle Methods](#5-lifecycle-methods)

- Before hooks were introduced, there are 2 reasons why you would use a class component instead of function component.
    - 1: To use state.
    - 2: To hook into different lifecycle methods of a component.

- Lifecycle component: any given component throughout its lifetime, goes into different phrases (Mounting, Updating, Unmounting) 
    - Mounting: when the component first mounts on to the screen
    - Updating: can happens multiple time as long as the component is on the screen; happens any time the state has changed and React has to update that component to reflect that change
    - Unmounting: when the component is removed from screen

- With lifecycle methods in class, we can hook different aspect of each of these phases and run code when necessary.
- Below are some of the lifecycle methods in each of the phases.

## [componentDidMount() Method](#componentdidmount-method)

- Whatever code you put inside this method is code that will run immediately after the very first render of the component.
- One of the things we might want to do as soon as our component mounts is to:
    - Make a call to API
    - Get data from db etc

- Notice that our state re-renders the second time, this is because after `componentDidMount()` function, the state has changed, so the component re-renders.

- since `componentDidMount()` method only run once as soon as the component renders, it is similar to React.useEffect with empty dependency array (which essentially will only run once as soon as the app renders).

```
React.useEffect(() => {
    // some code here
    }, [])
```

> 

## [componentDidUpdate() Method](#componentdidupdate-method)

- Will run everytime there is changes to either state or props to the component.
- Any time there is changes to the either state or props to a component, React will automatically re-render the component. Then, if we have defined `componentDidUpdate()` method, it will run whatever code inside that.
- Notice the `DidUpdate` in the function name -- it is past tense - ie the component did just recently update? then run `componentDidUpdate()`.

## [componentWillUnmount() Method](#componentwillunmount-method)

- Use this to clean up any side effects that we had existing before.
- Some examples on where we would use this are:
    - remove added event listener
    - disconnect any subscriptions we have set up before, eg web sockets, chat API that is listening for updates in a database, db connections etc.
- Helps clean up memory leak happening in the background.

# [6. Some other Lifecycle Methods](#6-some-other-lifecycle-methods)

- `shouldComponentUpdate()` method: Tell React whether or not the given component (the component where you write this lifecycle method), should update when there are state or props changes.
- `static getDerivedStateFromProps()` method: Take incoming props and create state that you derive from those props (ie derive state from props) in the component where you implement this lifecycle.
- `getSnapshotBeforeUpdate()`: Allow React to remember some information about the DOM (eg the scroll position) before re-render. Whatever data we save using this function will be passed to the `componentDidUpdate()` lifecycle method as a third parameter.
